## Atividade

- Leia uma imagem
- Crie variáveis para armazenar cada componente do RGB
- Salve cada imagem em disco

```python
import matplotlib.pyplot as plt
from skimage.io import imread, imsave

# 1. Ler a imagem 'personagem.png'
personagem = imread('personagem.png')

print(f"Dimensões da imagem personagem.png: {personagem.shape}")

# Exibir a imagem original
plt.figure(figsize=(8, 8))
plt.imshow(personagem)
plt.title('Imagem Original: personagem.png')
plt.axis('off')
plt.show()

# 2. Criar variáveis para armazenar cada componente do RGB
red_personagem = personagem[:,:,0]
green_personagem = personagem[:,:,1]
blue_personagem = personagem[:,:,2]

# Exibir os canais RGB
fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(red_personagem, cmap='gray')
ax[0].set_title('Canal Vermelho (Red)')
ax[0].axis('off')

ax[1].imshow(green_personagem, cmap='gray')
ax[1].set_title('Canal Verde (Green)')
ax[1].axis('off')

ax[2].imshow(blue_personagem, cmap='gray')
ax[2].set_title('Canal Azul (Blue)')
ax[2].axis('off')

plt.tight_layout()
plt.show()

# 3. Salvar cada imagem em disco
imsave('personagem_red.png', red_personagem)
imsave('personagem_green.png', green_personagem)
imsave('personagem_blue.png', blue_personagem)

print("Imagens dos canais RGB salvas como 'personagem_red.png', 'personagem_green.png' e 'personagem_blue.png'.")
```

## Atividade
- Percorra sua imagem e ache o pixel com maior valor em cada uma das componentes

```python
import numpy as np

# As variáveis red_personagem, green_personagem e blue_personagem já foram criadas
# na execução anterior e contêm os canais separados da imagem 'personagem.png'.

# Encontrar o maior valor e suas coordenadas para o canal Vermelho
max_red_value = np.max(red_personagem)
max_red_coords = np.unravel_index(np.argmax(red_personagem), red_personagem.shape)

# Encontrar o maior valor e suas coordenadas para o canal Verde
max_green_value = np.max(green_personagem)
max_green_coords = np.unravel_index(np.argmax(green_personagem), green_personagem.shape)

# Encontrar o maior valor e suas coordenadas para o canal Azul
max_blue_value = np.max(blue_personagem)
max_blue_coords = np.unravel_index(np.argmax(blue_personagem), blue_personagem.shape)

print(f"Maior valor no canal Vermelho: {max_red_value} no pixel {max_red_coords}")
print(f"Maior valor no canal Verde: {max_green_value} no pixel {max_green_coords}")
print(f"Maior valor no canal Azul: {max_blue_value} no pixel {max_blue_coords}")
```

# Atividade
- Faça uma binarização na imagem variando o limiar entre 0 e 1 com passos de 0.2
- Plot os resultados

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.color import rgb2gray

cinza = (rgb2gray(personagem[:, :, :3])*255).astype(np.uint8)

# A imagem 'cinza' está disponível do estado anterior do notebook (como uint8, 0-255).
# Para aplicar limiares entre 0 e 1, precisamos normalizá-la.
cinza_normalizada = cinza / 255.0

# Definir os limiares de 0 a 1 com passos de 0.2
limiares = np.arange(0.0, 1.2, 0.2) # Inclui 1.0

# Preparar o layout para os plots
num_plots = len(limiares) + 1 # Imagem original + imagens binarizadas
rows = 2
cols = int(np.ceil(num_plots / rows))

fig, axes = plt.subplots(rows, cols, figsize=(cols * 5, rows * 5))
axes = axes.flatten() # Achatar o array de eixos para fácil iteração

# Plotar a imagem original em escala de cinza
axes[0].imshow(cinza_normalizada, cmap='gray')
axes[0].set_title('Imagem Original (Grayscale)')
axes[0].axis('off')

# Iterar sobre os limiares e aplicar a binarização
for i, limiar in enumerate(limiares):
    # Binarização: pixels maiores que o limiar se tornam brancos (True), caso contrário pretos (False)
    imagem_binarizada = cinza_normalizada > limiar
    axes[i + 1].imshow(imagem_binarizada, cmap='gray')
    axes[i + 1].set_title(f'Limiar = {limiar:.1f}')
    axes[i + 1].axis('off')

# Ocultar quaisquer subplots não utilizados, se houver
for j in range(num_plots, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

## Atividade
- Transforme uma imagem para preto e branco
- Utilize outros métodos que não seja o de Sobel para encontrar as bordas. Ex: scharr, prewitt e roberts
- Faça um plot com todos os métodos que você utilizou

```python
import matplotlib.pyplot as plt
import numpy as np
from skimage.io import imread
from skimage.color import rgb2gray
from skimage.filters import scharr, prewitt, roberts

# 1. Carregar a imagem 'personagem.png'
personagem = imread('personagem.png')

# 2. Transformar a imagem para preto e branco (tons de cinza)
personagem_cinza = rgb2gray(personagem[:, :, :3])

# 3. Aplicar os métodos de detecção de bordas (Scharr, Prewitt, Roberts)
bordas_scharr = scharr(personagem_cinza)
bordas_prewitt = prewitt(personagem_cinza)
bordas_roberts = roberts(personagem_cinza)

# 4. Fazer um plot com todos os métodos utilizados
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(15, 12), sharex=True, sharey=True)
ax = axes.ravel()

ax[0].imshow(personagem_cinza, cmap='gray')
ax[0].set_title('Imagem Original (Grayscale)')
ax[0].axis('off')

ax[1].imshow(bordas_scharr, cmap='gray')
ax[1].set_title('Bordas (Scharr)')
ax[1].axis('off')

ax[2].imshow(bordas_prewitt, cmap='gray')
ax[2].set_title('Bordas (Prewitt)')
ax[2].axis('off')

ax[3].imshow(bordas_roberts, cmap='gray')
ax[3].set_title('Bordas (Roberts)')
ax[3].axis('off')

plt.tight_layout()
plt.show()
```
